```python
# Cambiar el formato de visualización de los números flotantes a dos decimales
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

```python
df['Price'].describe()
```

En cuanto al precio, se encuentra que tiene una media de $1075684.08 Una desviación estándar de $639310.72. El 25% de las propiedades tienen un precio de $650000 el percentil 50 es la mediana de los datos, con lo que se diferencia de la media

```python
print(df['Price'].quantile([0.05, 0.25, 0.5, 0.75, 0.95]))
```

```python
ax = (df['Price'].plot.box(figsize=(3,4)))

plt.tight_layout()
plt.show()
```

```python
ax = df['Rooms'].plot.box(figsize=(3,4))

plt.tight_layout()
plt.show()
```
